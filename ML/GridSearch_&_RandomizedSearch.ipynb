For Random Forest

```python
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.ensemble import RandomForestClassifier  # 1. Import Random Forest
from sklearn.model_selection import GridSearchCV

# 2. Load dataset (keeping your exact data pipeline)
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
daat_frame = pd.DataFrame(breast_cancer_dataset.data, columns=breast_cancer_dataset.feature_names)
daat_frame['label'] = breast_cancer_dataset.target

# Prepare X and y
X = daat_frame.drop(columns=['label'])
y = daat_frame['label']
X = np.array(X)
y = np.array(y)

# 3. Initialize Random Forest model (using your favorite random_state!)
model = RandomForestClassifier(random_state=21)

# 4. Define Random Forest hyperparameters to tune
parameter = {
    'n_estimators': [10, 50, 100, 200],      # Number of trees in the forest
    'max_depth': [None, 5, 10, 20],           # Maximum depth of the trees
    'min_samples_split': [2, 5, 10]           # Minimum samples required to split a node
}

# 5. Set up Grid Search
classifier = GridSearchCV(model, parameter, cv=5)

# 6. Fit the model
classifier.fit(X, y)

# 7. Print best results
print("Best Parameters:", classifier.best_params_)
print("Highest Accuracy:", classifier.best_score_)
```

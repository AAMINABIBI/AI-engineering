AND GATE by EPOCH WEIGHT UPDATE

````python
import numpy as np

X=np.array([[0,0],[0,1],[1,0],[1,1]])
y=np.array([0,0,0,1])

w=np.zeros(2)
b=0
lr=0.1

for epoch in range(10):
  for xi,yi in zip(X,y):
    pred =1 if xi@w +b >=0 else 0
    w=w+lr*(yi-pred)*xi
    b=b+lr*(yi-pred)

print('w=',w,'b=',b)
print([1 if xi @ w+b>0 else 0 for xi in X])
````

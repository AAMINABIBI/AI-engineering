```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset=pd.read_csv("Churn_Modelling (1).csv")
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
input_data=dataset.iloc[:,:-1]
output_data=dataset.iloc[:,-1]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss=StandardScaler()
ss.fit_transform(input_data)
#
```

```python
input_data=pd.DataFrame(ss.fit_transform(input_data)
,columns=input_data.columns)
#
```

```python
input_data.head()
```

ANN

```python
import tensorflow
```

```python
from keras.layers import Dense
from keras.models import Sequential
```

```python
ann=Sequential()
```

check Inputs


```python
input_data.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(input_data,output_data,test_size=0.2,random_state=42)
#
#
```

```python
ann.add(Dense(6,input_dim=8,activation="relu"))
ann.add(Dense(4,activation="relu"))
ann.add(Dense(2,activation="relu"))
ann.add(Dense(1,activation="sigmoid"))
```

```python
ann.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
ann.fit(x_train,y_train,batch_size=100,epochs=50)
```

```python
ann.predict(x_test)
```

```python
prd=ann.predict(x_test)
```

```python
prd_data=[]
for i in prd:
  if i>0.5:
    prd_data.append(1)
  else:
    prd_data.append(0)
```

```python
prd_data
```

```python
pred1=ann.predict(x_train)
prd_data1=[]
for i in pred1:
  if i>0.5:
    prd_data1.append(1)
  else:
    prd_data1.append(0)
```

TESTING ACCURACY

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,prd_data)*100
```

```python
accuracy_score(y_train,prd_data1)*100
```

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

tokenizer=T5Tokenizer.from_pretrained("t5-small")
model=T5ForConditionalGeneration.from_pretrained("t5-small")
#prepare input woth the task prefix

text="translate ennglish to french: Macine learning is interesting"
input_ids=tokenizer(text, return_tensors="pt").input_ids
#gemerate output token and decode it
outputs=model.generate(input_ids,max_new_tokens=30)
print(tokenizer.decode(outputs[0],skip_special_tokens=True))

```

```python
from transformers import BartTokenizer, BartForConditionalGeneration

tokenizer=BartTokenizer.from_pretrained("facebook/bart-large-cnn")
model=BartForConditionalGeneration.from_pretrained("facebook/bart-large-cnn")
#prepare input woth the task prefix

text="The lantern flickered against the stone wall of the abandoned lighthouse as Mara climbed the final spiral step Outside the sea hammered the cliffs like a restless heart She had come searching for her grandfathers lost journal convinced the old mans rumors of a voice in the waves were mere delirium At the top the lantern room was dry smelling faintly of salt and old cedar Resting on the brass pedestal was a leather bound ledger untouched by dust When Mara opened it the pages were blank until a sudden flash of lightning illuminated the glass dome Ink bloomed across the parchment like dark tidewater forming a single sentence You are finally late Before she could gasp the heavy iron trapdoor slammed shut beneath her feet and the beam of the dead lighthouse roared to life casting a violet glow over the empty ocean"
inputs=tokenizer(text,return_tensors="pt", max_length=1024,truncation=True)
#gemerate output token and decode it

summary_ids=model.generate(inputs["input_ids"],num_beams=4,min_length=20)
print(tokenizer.decode(summary_ids[0],skip_special_tokens=True))
```
